# Introduction : Synthèse de Nanocristaux Cœur-Coquille ($\text{InP/ZnS}$) pour la Nanothermométrie

Ce notebook est dédié au traitement des données expérimentales issues de l'étude optique de **boîtes quantiques (Quantum Dots)** semi-conductrices. Contrairement aux applications d'affichage classiques où l'on fait varier la taille des nanocristaux pour modifier l'émission, nous étudierons ici l'influence de la **température** sur leurs propriétés de photoluminescence, ouvrant la voie à des applications en **nanothermométrie**.

### Étape 1 : Nucléation et Croissance du Cœur ($\text{InP}$)
* **Réactifs :** Chlorure d'indium ($\text{InCl}_3$) comme précurseur d'indium, et Tris(diéthylamino)phosphine ($\text{P(NEt}_2)_3$) comme précurseur de phosphore.
* **Conditions opératoires :** * Température de réaction : **180°C** pendant **30 minutes**.
  * Atmosphère inerte : Sous **Argon ($\text{Ar}$)** pour éviter l'oxydation des précurseur et des nanocristaux.
  * Milieu : Présence de chlorure de zinc ($\text{ZnCl}_2$) et d'oléylamine (surfactant/solvant).
* **Produit obtenu :** Nanocristaux de Phosphure d'Indium ($\text{InP}$) qui forment le **cœur** (core) semi-conducteur.

### Étape 2 : Passivation par la Coquille ($\text{ZnS}$)
* **Réactifs :** Stéarate de zinc ($\text{Zn(stearate)}_2$) et Sulfure de trioctylphosphine ($\text{TOP-S}$), apportant respectivement les ions $\text{Zn}^{2+}$ et $\text{S}^{2-}$.
* **Conditions opératoires :** Chauffage plus intense entre **200°C et 260°C** pendant **3 heures**, toujours sous atmosphère inerte.
* **Produit final :** Hétérostructure **$\text{InP/ZnS}$ (Cœur-Coquille)**.

## 2. Intérêt Physique de la Structure Cœur-Coquille

À l'échelle nanométrique, le rapport surface/volume est extrêmement élevé. Un cœur d' $\text{InP}$ seul possède de nombreux défauts cristallins à sa surface (liaisons pendantes), appelés **pièges de surface**. Ces pièges induisent des transitions non-radiatives qui effondrent le rendement quantique de photoluminescence.

La croissance d'une coquille de **Sulfure de Zinc ($\text{ZnS}$)**, un semi-conducteur à grande bande interdite ($E_g \approx 3.6 \text{ eV}$), permet :
1. **La passivation de surface :** Elle sature les liaisons pendantes du cœur et élimine les défauts de surface.
2. **Le confinement des porteurs :** Elle crée une barrière de potentiel qui confine fortement la fonction d'onde de l'exciton (paire électron-trou) à l'intérieur du cœur d' $\text{InP}$ ($E_g \approx 1.35 \text{ eV}$).

> **Résultat :** Les recombinaisons radiatives sont hautement privilégiées, conférant aux Quantum Dots une intensité de photoluminescence stable et exploitable pour des mesures physiques de précision.
## 3. Phénomène de Photoluminescence : Excitation UV vs Émission Visible

La comparaison visuelle entre l'éclairage ambiant (solution incolore) et l'exposition sous rayonnement ultraviolet (luminescence colorée) illustre le processus de **photoluminescence par transition radiative**.  
  
<img src="figures/uv_vis.PNG" width="300" alt="UV_VIS">  
  
### Mécanisme Physique (Approche Énergétique)

Le phénomène se décompose en trois phases microphysiques distinctes :

1. **Absorption résonante :** Le rayonnement UV incident apporte des photons de forte énergie ($h\nu_{\text{excitation}}$). Si cette énergie est supérieure à la bande interdite (gap) du semi-conducteur ($h\nu > E_g$), un électron de la bande de valence est promu vers la bande de conduction, créant une paire électron-trou liée (état excité).
2. **Relaxation thermique (Non-radiative) :** L'électron et le trou relaxent très rapidement (en quelques picosecondes) vers les extréma de leurs bandes respectives (bords de bande) en perdant un peu d'énergie sous forme de vibrations du réseau cristallin (**phonons**).
3. **Recombinaison radiative (Émission) :** L'électron se recombine avec le trou en émettant un photon d'énergie $h\nu_{\text{émission}} \approx E_g$.

### Pourquoi passe-t-on de l'UV au Visible ?

Ce décalage en longueur d'onde est régi par deux principes physiques fondamentaux :

* **Le déplacement de Stokes (Stokes Shift) :** En raison des pertes d'énergie par relaxation thermique (phonons), l'énergie du photon émis est strictement inférieure à celle du photon absorbé :
$$\Delta E = h\nu_{\text{excitation}} - h\nu_{\text{émission}} > 0 \implies \lambda_{\text{émission}} > \lambda_{\text{excitation}}$$
C'est pourquoi une excitation dans l'**UV** ($\lambda \approx 365\text{ nm}$, haute énergie) donne lieu à une émission dans le **visible** ($\lambda \in [400 - 700]\text{ nm}$, plus basse énergie).


### Couplage Température - Énergie d'Émission : Le Principe de la Nanothermométrie

En nanothermométrie, on exploite le fait que les propriétés thermodynamiques du réseau cristallin influencent directement sa structure électronique. Lorsque la **température ($T$)** du milieu augmente, deux effets physiques conjugués se produisent :

1. **Dilatation thermique du réseau :** L'augmentation de la température accroît l'amplitude des vibrations atomiques, ce qui augmente la distance interatomique moyenne. Cela diminue le potentiel cristallin vu par les électrons et entraîne une **diminution de la bande interdite ($E_g$)**.
2. **Couplage électron-phonon :** L'interaction entre les porteurs de charge (excitons) et les vibrations du réseau (phonons) s'intensifie, ce qui renormalise les énergies des bandes de valence et de conduction, réduisant également $E_g$.

### Traduction Spectroscopique

Ces variations sont modélisées macroscopiquement par la **loi empirique de Varshni** :

$$E_g(T) = E_g(0) - \frac{\alpha T^2}{T + \beta}$$

*Où $\alpha$ et $\beta$ sont des constantes caractéristiques du matériau ($\text{InP}$).*

> **Conséquence directe pour le traitement de données :** > * Lorsque la température **augmente**, le gap $E_g$ diminue. L'énergie du photon émis ($h\nu_{\text{émission}}$) baisse, ce qui provoque un **décalage vers le rouge (redshift)** de la longueur d'onde d'émission $\lambda_{\text{émission}}$.
> * À l'inverse, un refroidissement provoque un **décalage vers le bleu (blueshift)**.

En mesurant précisément la position du pic de photoluminescence, on peut donc remonter à la température locale du nanocristal avec une résolution nanométrique.

## 4. Dispositif Expérimental et Protocole d'Étalonnage

Pour exploiter ces Quantum Dots en nanothermométrie, il est indispensable de réaliser préalablement un **étalonnage (calibration)** précis. Le but est de corréler quantitativement la température thermodynamique du milieu avec la réponse spectrale des nanocristaux.

<img src="figures/setup1.PNG" height="300" alt="Setup">
  
<img src="figures/setup2.PNG" height="300" alt="Setup">

### 4.1 Description du Banc de Mesure

Le dispositif expérimental est constitué de trois modules principaux :

* **Le bloc thermique (Contrôle et Consigne) :** La solution de Quantum Dots ($\text{InP/ZnS}$) est placée dans une cellule thermostatée contenant un **fil chauffant**. Ce système permet de balayer une gamme de température allant de **20°C à 200°C**. La température est pilotée et stabilisée en temps réel par une **régulation électronique** (de type PID).
* **La source d'excitation :** Une source lumineuse émettant dans l'**UV** irradie l'échantillon en continu pour générer l'état excité (génération d'excitons).
* **Le système de détection :** La lumière de photoluminescence émise par la solution est collectée, puis analysée par un **spectromètre** qui mesure l'intensité lumineuse en fonction de la longueur d'onde $\lambda$.

---

### 4.2 Protocole d'Acquisition et Cycle Thermique

Afin de s'assurer de la réversibilité et de la stabilité de la réponse des nanocristaux , le protocole est le suivant :

1. **Cycles en température :** On applique à la solution des cycles de montée et de descente en température via le régulateur entre 20°C et 200°C.
2. **Mesure simultanée (Synchronisation des données) :** À chaque mesure de température, une fois l'équilibre thermique atteint, on réalise une acquisition synchrone :
   * De la **température $T$** (mesurée par une sonde locale au cœur de la solution).
   * Du **spectre de photoluminescence $I = f(\lambda)$**.

> **Objectif du traitement de données :** À partir de cette matrice de spectres enregistrés à différentes températures, votre travail consistera à extraire la position du pic d'émission (par ajustement gaussien ou lorentzien) pour tracer la courbe d'étalonnage $\lambda_{\text{pic}} = f(T)$ et déterminer la sensibilité du nanothermomètre.


## 5. Prétraitement des Données et Extraction de Caractéristiques (*Feature Engineering*)

Pour chaque spectre d'émission $I = f(\lambda)$ enregistré à une température $T$ donnée, nous allons extraire des descripteurs géométriques et physiques. L'objectif est de réduire la dimensionnalité de chaque spectre (passer d'un vecteur de centaines de points d'intensité à un ensemble réduit de paramètres pertinents).

Pour chaque courbe, vous pourrez faire des codes permettant d'extraire les **6 caractéristiques (features)** suivantes :

1. **L'intensité maximale ($\mathit{I}_{\max}$) :** L'amplitude du signal au sommet du pic.
2. **La longueur d'onde au maximum ($\lambda_{\max}$) :** La position spectrale de ce sommet, directement liée à l'énergie du gap $E_g(T)$.
3. **La longueur d'onde de demi-intensité à gauche ($\lambda_{\text{half1}}$) :** La longueur d'onde du flanc gauche (côté des courtes longueurs d'onde / hautes énergies) pour laquelle $I = \mathit{I}_{\max} / 2$.
4. **La longueur d'onde de demi-intensité à droite ($\lambda_{\text{half2}}$) :** La longueur d'onde du flanc droit (côté des grandes longueurs d'onde / basses énergies) pour laquelle $I = \mathit{I}_{\max} / 2$.
5. **La largeur à mi-hauteur ($\text{FWHM}$) :** La largeur totale du pic à mi-hauteur, calculée par la différence :
$$\text{FWHM} = \lambda_{\text{half2}} - \lambda_{\text{half1}}$$
6. **Le ratio de dissymétrie des surfaces ($\text{Surface ratio}$) :** Le rapport entre l'aire sous la courbe de la partie droite (au-delà de $\lambda_{\max}$) et l'aire de la partie gauche (en deçà de $\lambda_{\max}$) :
$$\text{Surface ratio} = \frac{\int_{\lambda_{\max}}^{\infty} I(\lambda) d\lambda}{\int_{-\infty}^{\lambda_{\max}} I(\lambda) d\lambda}$$
*Remarque : Ce ratio permet de quantifier l'asymétrie du pic (due par exemple aux queues d'état ou aux contributions de phonons), qui peut évoluer avec la température.*

## 6. Perspective : Vers le Machine Learning

L'extraction de ces **6 caractéristiques** par spectre permet de construire une matrice de descripteurs de dimension $(N \times 6)$, où $N$ est le nombre total de spectres mesurés. 

Ce jeu de données épuré et hautement qualifié sera utilisé pour **entraîner des modèles de Machine Learning (régression)**. L'objectif final sera de prédire de manière robuste la température $T$ à partir de ces 6 descripteurs, en s'affranchissant des fluctuations de mesure ou des bruits instrumentaux qu'un étalonnage classique sur un seul paramètre ($\lambda_{\max}$ seul) pourrait subir.
